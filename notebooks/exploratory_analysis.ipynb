# Dataset Analysis and Synthetic Population: Summary of Findings

## 1. Dataset Overview

The project contains historical secondary-school data, higher-education data, and a semi-synthetic student population generated for simulating school dropout prevention through limited mentor interventions.

The datasets serve different purposes, including academic performance prediction, population generation, simulation initialization, and evaluation.

| Dataset | Records | Columns | Missing Values |
|---|---:|---:|---:|
| `student_dataset.csv` | 674 | 44 | 2,987 |
| `synthetic_student_population.csv` | 2,000 | 51 | 10,915 |
| `simulator_hidden_truth.csv` | 2,000 | 7 | 0 |
| `subject_records_reference.csv` | 1,044 | 39 | 0 |
| `student_outcomes_reference.csv` | 674 | 10 | 1,622 |
| `higher_ed_enrollment_features.csv` | 4,424 | 27 | 0 |
| `higher_ed_outcomes_reference.csv` | 4,424 | 14 | 0 |
| `variable_dictionary.csv` | 44 | 8 | 20 |

The datasets are not entirely independent. Several represent different views of the same student profiles and share identifiers or attributes.

## 2. Original Data Sources

### 2.1 Secondary-School Dataset

**Source:** UCI Student Performance Dataset, Portugal.

The original school data contains:

- 649 Portuguese course records.
- 395 Mathematics course records.
- 1,044 subject-level records in total.
- 358 conservatively identified one-to-one cross-subject links.

These records contain demographic, family, educational and academic performance information.

The processed `student_dataset.csv` contains **674 student profiles**, comprising 649 Portuguese anchors and 25 additional Mathematics-only profiles.

This is the primary data source used to construct the semi-synthetic student population.

**Limitation:** The original dataset measures academic performance, grades and absences. It does not contain verified school-dropout outcomes. Academic course failure must not be treated as equivalent to school dropout.

### 2.2 Higher-Education Dataset

**Source:** UCI Predict Students' Dropout and Academic Success.

The higher-education dataset contains 4,424 university student records with the following outcomes:

| Outcome | Students |
|---|---:|
| Dropout | 1,421 |
| Graduate | 2,209 |
| Enrolled | 794 |
| **Total** | **4,424** |

The observed dropout proportion is approximately 32.1% of all records, or 39.1% when considering only resolved Dropout and Graduate cases.

This dataset supports a separate university-dropout prediction experiment.

**The higher-education data was not directly used to generate the 2,000 synthetic school students.** The supplied files also do not establish whether it influenced undocumented modelling choices.

The school and higher-education datasets should remain separate because they represent different populations, educational contexts and outcome definitions.

## 3. Synthetic Student Population

### 3.1 Generation and Composition

The synthetic student population was created by resampling profiles from the processed secondary-school dataset and introducing additional generated attributes.

The generation process can be summarized as:

**Original Portuguese and Mathematics records → 674 processed school profiles → Bootstrap sampling and generated initial variables → 2,000 semi-synthetic students.**

Although the synthetic population contains 2,000 records, it represents only **615 distinct original student profiles**. Multiple synthetic students can therefore originate from the same historical profile.

The generated records are suitable as simulation agents but do not constitute 2,000 independently observed students.

### 3.2 Combined Synthetic Data Structure

Two files jointly represent the initial synthetic student state:

| File | Columns | Purpose |
|---|---:|---|
| `synthetic_student_population.csv` | 51 | Student characteristics, initial states and metadata |
| `simulator_hidden_truth.csv` | 7 | Hidden simulation parameters |
| **Combined dataset** | **57 unique** | Joined using `student_id` |

The total is 57 rather than 58 because `student_id` occurs in both files.

The resulting 57 columns consist of:

| Column category | Count |
|---|---:|
| Shared with processed school data | 44 |
| Additional synthetic population columns | 7 |
| Additional hidden simulation columns | 6 |
| **Total** | **57** |

The 44 shared columns are inherited from the processed school-student schema. Of these, 42 retain the values of the corresponding source profile; `student_id` and `record_provenance` change to represent the generated record.

### 3.3 Breakdown of the 44 Shared Columns

| Category | Columns | Description |
|---|---:|---|
| Identifiers and metadata | 9 | Student ID, linkage group, data split, original source and provenance |
| Historical school characteristics | 25 | Age, school, parental education, academic grades, past failures, internet availability and other contextual attributes |
| Derived disadvantage indicators | 5 | Indicators calculated from parental education, internet access and travel time |
| Unavailable information | 4 | Income, currency, verified economic disadvantage and observed school dropout |
| Assumed initial learning state | 1 | Generated initial learning-gap value |
| **Total** | **44** | |

Important fields include:

- `student_id`
- `school_id`
- `age_years`
- `sex_source`
- `mother_education_code`
- `father_education_code`
- `travel_time_category`

- `math_grade1_0_20`

- `disadvantage_proxy_group`

The disadvantage indicators are derived proxies, not verified evidence of household poverty or programme eligibility.

The unavailable fields `income_amount`, `income_currency`, `verified_economic_disadvantage` and `observed_school_dropout` contain missing values rather than measured information.

### 3.4 Additional Synthetic Population Attributes

Seven columns are added to the original 44-column schema:

| Column | Description |
|---|---|
| `source_anchor_student_id` | Historical profile from which the synthetic student originates |
| `initial_attendance_fraction_assumed` | Generated initial attendance |
| `initial_engagement_assumed` | Generated initial engagement |

| `support_sessions_completed` | Initialized intervention-session counter |
| `enrolled_initial_assumed` | Assumed initial enrollment status |
| `financial_need_verified` | Financial-need verification field, initially unavailable |

Attendance and engagement values are generated rather than directly observed in the original school records.

### 3.5 Hidden Simulation Parameters

The six additional non-identifier columns in `simulator_hidden_truth.csv` are:

| Column | Description |
|---|---|
| `annual_dropout_probability_assumed` | Assumed withdrawal probability over a reference period |
| `constant_weekly_hazard_assumed` | Derived weekly withdrawal hazard |
| `response_tendency_assumed` | Hidden intervention-response tendency |
| `financial_barrier_latent_assumed` | Generated financial-barrier indicator |
| `dropout_draw_at_static_annual_hazard_assumed` | Static-hazard diagnostic outcome draw |
| `provenance` | Parameter provenance information |

The assumed dropout probability has an approximate population mean of 9.95%. This is a **simulation assumption, not an empirical school-dropout rate**.

The static dropout draw must not predetermine outcomes in the future sequential intervention simulator.

## 4. Column Overlap and Dataset Relationships

Several datasets share columns because they reference the same original school profiles.

| Dataset | Columns shared with `student_dataset.csv` | Matching columns |
|---|---:|---|
| `synthetic_student_population.csv` | 44 | All 44 processed student column names |
| `split_manifest.csv` | 4 | `student_id`, `linkage_group_id`, `split`, `anchor_subject` |
| `subject_records_reference.csv` | 4 | `guardian`, `linkage_group_id`, `source_row`, `split` |
| `student_outcomes_reference.csv` | 3 | `student_id`, `anchor_subject`, `observed_school_dropout` |
| `higher_ed_enrollment_features.csv` | 2 | `linkage_group_id`, `split` |
| `higher_ed_outcomes_reference.csv` | 0 | None |

Matching column names do not necessarily imply matching students or equivalent field meanings.

### Student Identifier Relationships

- The 674 `student_id` values in the processed student, split-manifest and student-outcome tables correspond directly.
- The original subject records use `record_id`. All 674 processed anchor IDs are represented among these source records.
- Synthetic students receive independent `SYN...` identifiers and retain their historical parent profile in `source_anchor_student_id`.
- Hidden simulator records use the same 2,000 synthetic `student_id` values.
- Higher-education records use a separate `HE...` identifier namespace.

The higher-education records cannot be joined to school students on the basis of shared student identity.

The school `linkage_group_id` identifies conservative record-linkage clusters, not verified administrative identities.

## 5. Dataset Integration Recommendations

The datasets should be combined selectively according to their intended use.

| Combination | Join key | Purpose |
|---|---|---|
| Processed students + school outcomes | `student_id` | Historical academic performance prediction and evaluation |
| Processed students + split manifest | `student_id` | Train/validation/test integrity checking |
| Synthetic population + hidden simulator states | `student_id` | Simulation initialization, accessible to the environment only |
| Synthetic population + processed source profiles | `source_anchor_student_id` ↔ `student_id` | Source lineage and inheritance validation |
| Higher-education enrollment + outcomes | `record_id` | Independent university-dropout prediction |

The complete dataset should not be merged into one universal training table.

In particular, hidden simulator parameters and future outcome fields must remain inaccessible to ordinary allocation policies and prediction models at decision time.

## 6. Data Quality and Research Limitations

The audit established several important findings:

1. **Student identity consistency:** Matching source and processed student identifiers are consistent across the relevant datasets.
2. **Synthetic inheritance:** Historical student characteristics are preserved in generated records as intended.
3. **Training split integrity:** Synthetic students retain source-linked split assignments, preventing related student profiles from crossing train, validation and test partitions.
4. **Missing information:** Important socioeconomic attributes, including verified income and poverty status, are unavailable. Missing values should not automatically be replaced with zero or false.
5. **No observed school-dropout target:** Historical school data provides academic outcomes, not verified permanent withdrawal.
6. **No observed causal intervention effects:** Mentoring effectiveness, response probabilities and sequential dropout transitions cannot be identified from the supplied school data.
7. **No empirical mentor population:** Mentor skills, availability, caseload limits and costs must be introduced as explicitly synthetic scenario attributes.
8. **Limited generalizability:** The synthetic population inherits characteristics from two Portuguese schools, including an original age range of 15–22. It should not be assumed representative of disadvantaged schoolchildren in other countries.

## 7. Recommended Usage in the Research System

The project should distinguish historical data, simulation initialization, hidden environment states and controller observations.

| Research stage | Recommended data |
|---|---|
| Data exploration and preprocessing | Original subject records, variable dictionary and processed school data |
| Academic prediction baseline | Processed student profiles and historical course outcomes |
| Synthetic population initialization | `synthetic_student_population.csv` |
| Environment initialization | Synthetic population and hidden simulator parameters |
| Mentor allocation | Estimated student information and generated mentor resource data |
| Intervention simulation | Explicit stochastic state-transition and response models |
| LLM information extraction | Newly generated mentor notes based on observable events |
| State estimation | Validated extracted observations and prior estimated student states |
| Fairness and outcome evaluation | Simulation trajectories, resource logs and evaluator-only outcomes |
| Independent university-dropout benchmark | Higher-education enrollment and outcome datasets |

The intervention simulator should maintain separate **true environment states** and **estimated controller states**.

The environment determines simulated intervention responses. Mentor notes provide partial observations, an LLM extracts supported information, an independent state estimator updates student estimates, and a prediction model recalculates risk for the next allocation decision.

## 8. Conclusion

The dataset audit confirms that the project's semi-synthetic student population is derived primarily from the UCI Portuguese secondary-school performance data, with additional generated variables and assumed simulator parameters. The higher-education data constitutes a separate dataset and was not directly incorporated into the synthetic school population.

The combined synthetic student and hidden-state files provide **2,000 simulation agents described by 57 unique columns**. Their historical characteristics are grounded in observed school records, but their attendance, engagement, withdrawal hazards and intervention-response parameters are synthetic or assumed.

The existing data is suitable for initializing a reproducible, semi-synthetic study of resource-constrained mentor allocation. However, it does not establish real school-dropout rates or the causal effectiveness of educational interventions.

**The appropriate research objective is therefore to evaluate how different allocation policies and information-quality conditions affect simulated retention, fairness and resource utilization under transparent, explicitly documented assumptions.**

Real-world claims about dropout prevention would require additional longitudinal school records and credible intervention-outcome evidence.
